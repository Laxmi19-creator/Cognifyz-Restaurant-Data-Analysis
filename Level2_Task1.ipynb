```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Dataset .csv")
print(df.columns)
```

```python
print(df["Aggregate rating"].describe())
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Aggregate rating"], bins=10, edgecolor="black")

plt.title("Distribution of Aggregate Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")

plt.savefig("Restaurant_Rating_Distribution.png")

plt.show()
```

```python
rating_ranges = pd.cut(
    df["Aggregate rating"],
    bins=[0, 1, 2, 3, 4, 5],
    labels=["0-1", "1-2", "2-3", "3-4", "4-5"]
)

rating_count = rating_ranges.value_counts().sort_index()

print(rating_count)
most_common = rating_count.idxmax()

print("Most Common Rating Range:", most_common)
```

```python
average_votes = df["Votes"].mean()

print("Average Votes:", round(average_votes,2))
plt.figure(figsize=(8,5))

rating_count.plot(kind="bar")

plt.title("Restaurant Rating Ranges")
plt.xlabel("Rating Range")
plt.ylabel("Number of Restaurants")

plt.savefig("Rating_Ranges.png")

plt.show()
```
