```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Dataset .csv")
```

```python
highest_votes = df.loc[df["Votes"].idxmax()]

print("Restaurant with Highest Votes:")
print(highest_votes[["Restaurant Name", "Votes", "Aggregate rating"]])

lowest_votes = df.loc[df["Votes"].idxmin()]

print("Restaurant with Lowest Votes:")
print(lowest_votes[["Restaurant Name", "Votes", "Aggregate rating"]])
```

```python
top10_votes = df.nlargest(10, "Votes")

print(top10_votes[["Restaurant Name", "Votes", "Aggregate rating"]])
```

```python
correlation = df["Votes"].corr(df["Aggregate rating"])

print("Correlation between Votes and Ratings:", round(correlation, 2))
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df["Votes"], df["Aggregate rating"])

plt.title("Votes vs Aggregate Rating")
plt.xlabel("Votes")
plt.ylabel("Aggregate Rating")

plt.grid(True)

plt.savefig("Votes_vs_Rating.png")

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.bar(top10_votes["Restaurant Name"], top10_votes["Votes"])

plt.xticks(rotation=45, ha="right")

plt.title("Top 10 Most Voted Restaurants")
plt.xlabel("Restaurant Name")
plt.ylabel("Votes")

plt.tight_layout()

plt.savefig("Top10_Most_Voted_Restaurants.png")

plt.show()
```
