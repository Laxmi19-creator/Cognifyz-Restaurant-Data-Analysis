```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Dataset .csv")
print(df.columns)
highest_votes = df.loc[df["Votes"].idxmax()]

print("Restaurant with Highest Votes")
print(highest_votes[["Restaurant Name", "Votes", "Aggregate rating"]])
```

```python
lowest_votes = df.loc[df["Votes"].idxmin()]

print("Restaurant with Lowest Votes")
print(lowest_votes[["Restaurant Name", "Votes", "Aggregate rating"]])
```

```python
average_votes = df["Votes"].mean()

print("Average Votes:", round(average_votes,2))
```

```python
correlation = df["Votes"].corr(df["Aggregate rating"])

print("Correlation:", round(correlation,3))
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df["Votes"], df["Aggregate rating"])

plt.title("Votes vs Aggregate Rating")
plt.xlabel("Votes")
plt.ylabel("Aggregate Rating")

plt.savefig("Votes_vs_Rating.png")

plt.show()
```

```python
top10 = df.nlargest(10, "Votes")

print(top10[["Restaurant Name","Votes","Aggregate rating"]])
```

```python
plt.figure(figsize=(12,6))

plt.bar(top10["Restaurant Name"], top10["Votes"])

plt.xticks(rotation=45, ha="right")

plt.title("Top 10 Most Voted Restaurants")
plt.xlabel("Restaurant")
plt.ylabel("Votes")

plt.tight_layout()

plt.savefig("Top10_Voted_Restaurants.png")

plt.show()
```
