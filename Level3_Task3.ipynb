```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Dataset .csv")
print(df.columns)
online_delivery = pd.crosstab(
    df["Price range"],
    df["Has Online delivery"]
)

print(online_delivery)
```

```python
online_percentage = pd.crosstab(
    df["Price range"],
    df["Has Online delivery"],
    normalize="index"
) * 100

print(online_percentage)
```

```python
online_percentage.plot(kind="bar", figsize=(8,5))

plt.title("Price Range vs Online Delivery")
plt.xlabel("Price Range")
plt.ylabel("Percentage")

plt.tight_layout()
plt.savefig("Price_vs_OnlineDelivery.png")
plt.show()
```
