```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Dataset .csv")
print(df.columns)
```

```python
city_counts = df["City"].value_counts()

print(city_counts)
```

```python
top_city = city_counts.idxmax()
restaurant_count = city_counts.max()

print("City with Highest Restaurants:", top_city)
print("Number of Restaurants:", restaurant_count)
```

```python
average_rating = df.groupby("City")["Aggregate rating"].mean()

#print(average_rating)
average_rating = average_rating.round(2)
print(average_rating)
```

```python
highest_city = average_rating.idxmax()
highest_rating = average_rating.max()

print("Highest Rated City:", highest_city)
print("Average Rating:", round(highest_rating,2))
```

```python
top10 = city_counts.head(10)

plt.figure(figsize=(10,5))
top10.plot(kind="bar")

plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.savefig("Top_10_Cities.png")
plt.show()
```
