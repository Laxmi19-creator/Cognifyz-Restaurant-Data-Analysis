```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Dataset .csv")
```

```python
#total_restaurants = len(df)
total_restaurants = len(df)
print(total_restaurants)
```

```python
# Split cuisines
cuisines = df["Cuisines"].str.split(", ")

# Convert into separate rows
all_cuisines = cuisines.explode()

# Count cuisines
top_cuisines = all_cuisines.value_counts()

# Top 3 cuisines
top3 = top_cuisines.head(3)

print(top3)
```

```python
percentage = (top3 / total_restaurants) * 100

print(percentage)
```

```python
plt.figure(figsize=(8,5))

top3.plot(kind="bar", color=["blue","green","orange"])

plt.title("Top 3 Most Common Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")

plt.savefig("Top_3_Cuisines.png")
plt.show()
```
