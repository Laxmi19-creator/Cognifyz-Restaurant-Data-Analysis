```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Dataset .csv")
print(df.columns)
```

```python
price_count = df["Price range"].value_counts().sort_index()

print(price_count)
```

```python
total_restaurants = len(df)

percentage = (price_count / total_restaurants) * 100

print(percentage)
```

```python
for price, count in price_count.items():
    print(f"Price Range {price}: {count} restaurants ({percentage[price]:.2f}%)")
```

```python
plt.figure(figsize=(8,5))

price_count.plot(kind="bar")

plt.title("Distribution of Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")

plt.savefig("Price_Range_Distribution.png")

plt.show()
```
